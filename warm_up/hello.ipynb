```python
cat_img = imread("grumpy_cat.webp")
```

```python
print(type(cat_img))
print(cat_img.shape)
```

```python
plt.title("Somurtkan Kedi")
plt.imshow(cat_img)
plt.show()
```

```python
r_channel = cat_img[:, :, 0]
g_channel = cat_img[:, :, 1]
b_channel = cat_img[:, :, 2]
print(r_channel.shape, g_channel.shape, b_channel.shape)
```

```python
zeros = np.zeros_like(r_channel)
print(f"zeros shape = {zeros.shape}")

red_img = np.stack([r_channel, zeros, zeros], axis=2)
print(f"red_img shape = {red_img.shape}")

green_img = np.stack([zeros, g_channel, zeros], axis=2)
blue_img = np.stack([zeros, zeros, b_channel], axis=2)

plt.figure(figsize=(4, 8))

plt.subplot(3, 1, 1)
plt.imshow(red_img)
plt.title("Red Only")
plt.axis("off")

plt.subplot(3, 1, 2)
plt.imshow(green_img)
plt.title("Green Only")
plt.axis("off")

plt.subplot(3, 1, 3)
plt.imshow(blue_img)
plt.title("Blue Only")
plt.axis("off")

plt.show()
```
